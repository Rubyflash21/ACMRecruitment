```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv("cleaned_students_performance.csv")

df.head()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="math score", bins=10, kde=True)

plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")

plt.show()
```

```python
average_scores = df[["math score", "reading score", "writing score"]].mean()

average_scores
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=average_scores.index,
    y=average_scores.values,
    hue=average_scores.index,
    palette="viridis",
    legend=False
)

plt.title("Average Student Score by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df[["math score", "reading score", "writing score"]],
    palette="Set2"
)

plt.title("Score Distribution by Subject")
plt.xlabel("Subject")
plt.ylabel("Score")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="reading score",
    y="writing score",
    alpha=0.6
)

plt.title("Relationship Between Reading and Writing Scores")
plt.xlabel("Reading Score")
plt.ylabel("Writing Score")

plt.show()
```

```python
df["reading score"].corr(df["writing score"])
```

```python
correlation_matrix = df[
    ["math score", "reading score", "writing score"]
].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between Student Scores")

plt.show()
```
